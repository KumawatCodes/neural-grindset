## LinkedList Implementation

```python
# Node class
class Node:
    def __init__(self,data):
        self.data = data
        self.next = None
```

```python
# List class
class Linkedlist:
    def __init__(self):
        self.head = None


    # append at head
    def appendAtHead(self,data):
        new_node = Node(data);
        new_node.next = self.head
        self.head = new_node
    # append at tail
    def appendAtTail(self,data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
    
    # display list
    def display(self):
        temp = self.head
        while temp:
            print(temp.data,end="->")
            temp= temp.next
        print("None")
    
```

```python
ll = Linkedlist()

ll.appendAtHead(30)
ll.appendAtHead(20)
ll.display()
ll.appendAtHead(10)
ll.display()
```
