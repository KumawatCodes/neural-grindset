````python
# ============================================================
# SUB-TOPIC: Node Class — The Atomic Unit of a Linked List
# ============================================================

# --- THEORY ---
# What: A Node is a container that holds:
#   1. data  — the actual value you want to store
#   2. next  — a reference (pointer) to the NEXT node in the list
#
# Why: Without Node, we'd have no way to "chain" elements together.
#      Every linked list operation starts by creating or traversing Nodes.
#
# BEGINNER: Think of Node as a box with two compartments:
#   - Left compartment: your actual data (a number, string, anything)
#   - Right compartment: a sticky note with the address of the next box
#
# ADVANCED: In Python, `self.next = None` doesn't pre-allocate memory
#   for a node — it's just a null reference. The actual memory for the next
#   Node is allocated when you do `Node(value)`. Python uses reference counting
#   + cyclic garbage collection. When no variable holds a reference to a node,
#   Python automatically reclaims that memory.
#
# Time Complexity: O(1) — constant time to create one node
# Space Complexity: O(1) — one node takes constant space
#                   O(n) total — n nodes for a list of n elements
# Edge Cases: None — Node creation never fails (unless system is out of memory)

# --- CODE ---
from __future__ import annotations  # Allows type hints to reference the class itself
from typing import Optional

class Node:
    """
    A single node in a singly linked list.
    
    Attributes:
        data: The value stored in this node (any type)
        next: Reference to the next node, or None if this is the last node
    
    Example:
        node1 = Node(10)   # Creates: [10 | None]
        node2 = Node(20)   # Creates: [20 | None]
        node1.next = node2 # Creates: [10 | *] → [20 | None]
    """
    def __init__(self, data: int) -> None:
        self.data: int = data
        self.next: Optional[Node] = None  # Always None when first created
    
    def __repr__(self) -> str:
        """String representation for debugging."""
        return f"Node({self.data})"


# --- DRY RUN TRACE ---
# 
# Step 1: node1 = Node(10)
#   Memory: Allocate space for a Node object
#   node1.data = 10
#   node1.next = None
#   State: node1 → [10 | None]
#
# Step 2: node2 = Node(20)
#   node2.data = 20
#   node2.next = None
#   State: node2 → [20 | None]
#
# Step 3: node1.next = node2
#   node1.next now POINTS TO the node2 object
#   State: node1 → [10 | *──────────► node2 → [20 | None]
#
# Step 4: node1.next.data
#   "Follow the pointer in node1.next, then read .data"
#   = node2.data = 20

# --- TEST ---
def test_node():
    # Test basic creation
    node = Node(42)
    assert node.data == 42
    assert node.next is None, "Newly created node should point to None"
    
    # Test linking two nodes
    node1 = Node(10)
    node2 = Node(20)
    node1.next = node2
    
    assert node1.next is node2, "node1.next should reference node2"
    assert node1.next.data == 20, "Following next pointer should give node2's data"
    assert node2.next is None, "node2 has no next"
    
    # Test chaining three nodes
    node3 = Node(30)
    node2.next = node3
    assert node1.next.next.data == 30, "Two hops: node1 → node2 → node3"
    
    print("✅ Node class tests passed!")

test_node()
````

## Cell 1 — The Pain Point, Analogy & Memory Layout

### 🔴 The Pain Point — Why did Linked Lists get invented?

Imagine you're storing 1 million exam scores in an array. You need to insert a new score at position 0 (the front). What happens?

**With an Array:**
```
Before: [89, 76, 92, 88, ...]  (1,000,000 elements)
Insert 95 at index 0:
  → Shift ALL 1,000,000 elements one position right: O(n) time
After:  [95, 89, 76, 92, 88, ...]
```

This is **painful**. Even deleting from the front costs O(n) because every element must shift left.

**The root problem:** Arrays store data in contiguous (adjacent) memory blocks. To insert/delete in the middle or front, you MUST physically move everything.

**The real-world trigger:** In early operating systems, managing process queues (which process runs next) required constant insertions and deletions at arbitrary positions. Doing this with arrays was catastrophically slow.

**Linked Lists solved this:** Each element (node) lives anywhere in memory and holds a pointer to the next one. Insertion/deletion = just rewire pointers. No shifting needed.

---

### 🟡 The Analogy — Explain to a 5-Year-Old

**Think of a treasure hunt.**

Each clue is written on a piece of paper. The clue doesn't tell you all the locations at once — it just tells you **where the NEXT clue is hidden**.

```
[Clue 1: "Under your bed"] ──points to──► [Clue 2: "In the kitchen"] ──points to──► [Clue 3: "In the garden"] ──► None (END)
```

- Each **clue paper** = a **Node** (holds data + address of next)
- The **treasure hunt master** who knows where Clue 1 is = the **head pointer**
- "None" at the end = **null pointer** (no more clues)

**Key insight:** You don't need the clues to be physically next to each other. Clue 1 can be in your bedroom, Clue 2 in the kitchen, Clue 3 in the garden — they're linked by instructions, not location.

---

### 🟢 Memory Layout — What Actually Happens in RAM

**Array `[10, 20, 30]`:** Elements are stored side-by-side in RAM
```
Memory Address:  1000   1004   1008
                ┌────┐ ┌────┐ ┌────┐
                │ 10 │ │ 20 │ │ 30 │
                └────┘ └────┘ └────┘
                  ^
                  arr[0] lives here. arr[1] = arr[0] + 4 bytes (calculated!)
```

**Linked List `10 → 20 → 30`:** Nodes can live ANYWHERE in RAM
```
Memory Address:  1000        3048        7200
                ┌────┬────┐ ┌────┬────┐ ┌────┬──────┐
                │ 10 │3048│ │ 20 │7200│ │ 30 │ None │
                └────┴────┘ └────┴────┘ └────┴──────┘
                  data  ^      data  ^     data   ^
                        │            │             │
                      "next"       "next"        "next"
                    pointer to    pointer to    (end of list)
                     node 2        node 3

head ──────────────► 1000 (points to first node)
```

**What a Node stores:**
```
┌──────────────────────────┐
│  Node Object             │
│  ┌──────────┬──────────┐ │
│  │  data    │  next    │ │
│  │  (value) │ (address)│ │
│  └──────────┴──────────┘ │
└──────────────────────────┘
```

**BEGINNER TIP:** `next` doesn't store the actual next node. It stores the **memory address** where the next node lives. Following a pointer = "go look at that address."

**ADVANCED NOTE:** In Python, everything is a reference. `self.next = new_node` doesn't copy the node — it stores a reference (essentially a pointer) to the object. Python's garbage collector automatically frees memory when no references point to a node.

## Cell 3 — LinkedList Base Class + add_at_head


````python
# ============================================================
# SUB-TOPIC: LinkedList Base + add_at_head
# Operation: Insert a new node at the FRONT of the list
# ============================================================

# --- THEORY ---
# What: Creates the LinkedList container and implements front insertion.
#
# Why add_at_head?
#   - It's O(1) — the fastest insertion possible for a linked list
#   - Useful for: implementing stacks, reversing a list by re-insertion,
#     recent-items lists (newest at front), undo history
#
# The key insight: We have DIRECT access to the head. New node just needs
# to point to the old head, then we update head to the new node.
#
# BEFORE add_at_head(5):
#   head → [10] → [20] → [30] → None
#
# AFTER add_at_head(5):
#   head → [5] → [10] → [20] → [30] → None
#
# Time Complexity: O(1) — always 3 operations regardless of list size
# Space Complexity: O(1) — allocate exactly 1 new node
# Edge Cases:
#   - Empty list: new node becomes both head and only element
#   - Any size list: same O(1) logic applies

# --- CODE ---
class LinkedList:
    """
    Singly Linked List implementation.
    
    We maintain a `head` pointer (the entry point) and a `size` counter
    (so get_size() is O(1) instead of O(n)).
    
    BEGINNER: The LinkedList class is like the "manager" of all nodes.
    It doesn't store data itself — it just keeps track of where the first
    node is (head) and how many nodes exist (size).
    
    ADVANCED: The `size` attribute is a space-time tradeoff. We spend O(1)
    extra space to track size so we can answer "how many elements?" in O(1)
    instead of O(n). This is the same principle as a length field in C strings
    vs. null-terminated strings.
    """
    
    def __init__(self) -> None:
        self.head: Optional[Node] = None
        self.size: int = 0
    
    def add_at_head(self, data: int) -> None:
        """
        Insert a new node at the beginning of the list.
        
        Algorithm:
            1. Create a new Node with the given data
            2. Make new node's .next point to the current head
            3. Update head to point to the new node
            4. Increment size
        
        Time: O(1) | Space: O(1)
        
        BEGINNER MISTAKE: Don't do step 3 before step 2!
        If you set head = new_node FIRST, you lose the reference to the
        old head and can't link them together.
        """
        new_node = Node(data)
        new_node.next = self.head   # Step 2: link new → old_head
        self.head = new_node        # Step 3: update head to new
        self.size += 1

    def display(self) -> list[int]:
        """
        Traverse list and return all values as a Python list.
        
        Time: O(n) — must visit every node
        Space: O(n) — result list stores all values
        
        BEGINNER: 'current' is our "finger" that walks through the list.
        We start at head and keep moving to .next until we hit None.
        """
        result = []
        current = self.head
        while current:              # While current is not None
            result.append(current.data)
            current = current.next  # Move finger to next node
        return result


# --- DRY RUN TRACE ---
#
# ll = LinkedList()
#   head = None, size = 0
#   State: (empty)
#
# ll.add_at_head(30)
#   new_node = Node(30)        → [30 | None]
#   new_node.next = self.head  → new_node.next = None (head was None)
#   self.head = new_node       → head now points to [30 | None]
#   size = 1
#   State: head → [30 | None]
#
# ll.add_at_head(20)
#   new_node = Node(20)        → [20 | None]
#   new_node.next = self.head  → new_node.next = [30 | None]  ← KEY STEP
#   self.head = new_node       → head points to new [20 | *]
#   size = 2
#   State: head → [20 | *] → [30 | None]
#
# ll.add_at_head(10)
#   new_node = Node(10)
#   new_node.next = [20 | *]   ← links to old head
#   head = [10 | *]
#   size = 3
#   State: head → [10 | *] → [20 | *] → [30 | None]

# --- TEST ---
def test_add_at_head():
    ll = LinkedList()
    
    # Test on empty list
    ll.add_at_head(30)
    assert ll.display() == [30], f"Expected [30], got {ll.display()}"
    assert ll.size == 1
    
    # Test sequential insertions (newest element always at front)
    ll.add_at_head(20)
    assert ll.display() == [20, 30]
    
    ll.add_at_head(10)
    assert ll.display() == [10, 20, 30]
    assert ll.size == 3
    
    # Verify head is always the last inserted element
    assert ll.head.data == 10
    
    print("✅ add_at_head tests passed!")

test_add_at_head()
````

## Cell 4 — add_at_tail


````python
# ============================================================
# SUB-TOPIC: add_at_tail
# Operation: Insert a new node at the END of the list
# ============================================================

# --- THEORY ---
# What: Appends a new node after the last existing node.
#
# Why: When order matters and you want FIFO (First In First Out) behavior,
#      you add at tail and remove at head. This is how queues work.
#      Also used for: building a list in order, appending logs, etc.
#
# The challenge: We have no direct reference to the last node (unlike head).
# We MUST walk the entire list to find the node whose .next is None.
#
# BEFORE add_at_tail(40):
#   head → [10] → [20] → [30] → None
#
# AFTER add_at_tail(40):
#   head → [10] → [20] → [30] → [40] → None
#
# Time Complexity: O(n) — must traverse to the last node
#   (ADVANCED: If you maintain a `tail` pointer, this becomes O(1)!
#    Trade-off: tail pointer costs O(1) space but must be carefully
#    maintained on every insert and delete operation.)
# Space Complexity: O(1) — only one new node
# Edge Cases:
#   - Empty list: new node becomes the head (special case!)
#   - List of 1: current.next is None immediately — still works

    def add_at_tail(self, data: int) -> None:
        """
        Append a new node at the end of the list.
        
        Algorithm:
            1. Create new node
            2. If list is empty: new node IS the head (special case)
            3. Otherwise: walk to last node (where .next is None)
            4. Link last node's .next to new node
            5. Increment size
        
        Time: O(n) | Space: O(1)
        
        BEGINNER TRAP: The empty list is a special case because self.head
        is None. If you try `current = self.head; current.next = new_node`
        on an empty list, you get AttributeError: 'NoneType' has no .next
        
        ADVANCED: The `while current.next` condition is crucial. It stops
        at the SECOND-TO-LAST position so we can do current.next = new_node.
        Using `while current` would stop one step too late (at None).
        """
        new_node = Node(data)
        
        if not self.head:               # Edge case: empty list
            self.head = new_node
        else:
            current = self.head
            while current.next:         # Walk until last node
                current = current.next
            current.next = new_node     # Link last node to new node
        
        self.size += 1


# --- DRY RUN TRACE ---
#
# Starting: head → [10 | *] → [20 | *] → [30 | None], size = 3
#
# add_at_tail(40):
#   new_node = Node(40) → [40 | None]
#   self.head is not None → go to else branch
#   current = self.head → current = [10 | *]
#
#   Iteration 1: current.next = [20 | *] (truthy) → current = [20 | *]
#   Iteration 2: current.next = [30 | None] (truthy) → current = [30 | None]
#   Iteration 3: current.next = None (falsy) → EXIT while loop
#
#   current is now [30 | None] (the last node)
#   current.next = new_node → [30 | *] → [40 | None]
#   size = 4
#
# Final: head → [10] → [20] → [30] → [40] → None

# --- TEST ---
def test_add_at_tail():
    ll = LinkedList()
    
    # Test on empty list
    ll.add_at_tail(10)
    assert ll.display() == [10]
    assert ll.head.data == 10
    
    # Test sequential — order should be preserved
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    assert ll.display() == [10, 20, 30]
    assert ll.size == 3
    
    # Verify head is still the first inserted
    assert ll.head.data == 10
    
    print("✅ add_at_tail tests passed!")
````

## Cell 5 — add_at_position


````python
# ============================================================
# SUB-TOPIC: add_at_position
# Operation: Insert a node at a specific index (0-based)
# ============================================================

# --- THEORY ---
# What: Insert a node at any arbitrary position in the list.
#
# Why: add_at_head and add_at_tail are special cases of this.
#      Real use cases: inserting a task at a specific priority position,
#      maintaining a sorted order, etc.
#
# Key Insight: To insert at position k, you need to:
#   1. Walk to position k-1 (the node BEFORE the target)
#   2. New node's .next = that node's current .next
#   3. That node's .next = new node
#
# BEFORE add_at_position(1, 15):
#   head → [10] → [20] → [30] → None
#            ^
#          index 0  1      2
#
# AFTER:
#   head → [10] → [15] → [20] → [30] → None
#
# Time Complexity: O(n) — worst case walk to position n-1
# Space Complexity: O(1) — one new node
# Edge Cases:
#   - position == 0: same as add_at_head
#   - position >= size: same as add_at_tail (or raise error)
#   - negative position: invalid input

    def add_at_position(self, position: int, data: int) -> None:
        """
        Insert node at given 0-based position.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: The trick is stopping ONE BEFORE the target position.
        If I want to insert at index 2, I walk to index 1 (the "previous"
        node) and rewire its .next.
        
        ADVANCED: This is why doubly linked lists exist — with DLL, you
        can also traverse backward, halving your walk time for positions
        in the second half of the list (still O(n) worst case, but better
        constant factor).
        """
        if position < 0:
            raise ValueError(f"Position must be non-negative, got {position}")
        
        if position == 0:
            self.add_at_head(data)
            return
        
        if position >= self.size:
            self.add_at_tail(data)
            return
        
        new_node = Node(data)
        current = self.head
        
        # Walk to position - 1
        for _ in range(position - 1):
            current = current.next
        
        # Insert: new_node → (what was at position) → rest
        new_node.next = current.next   # Step 1: link new → next
        current.next = new_node        # Step 2: link prev → new
        self.size += 1


# --- DRY RUN TRACE ---
#
# List: head → [10] → [20] → [30] → None, size = 3
# add_at_position(1, 15)
#
# position = 1, not 0, not >= size
# new_node = Node(15) → [15 | None]
# current = head → [10 | *]
#
# Loop range(0) (position-1 = 0): ZERO iterations, current stays at [10]
#
# new_node.next = current.next → [15 | *] → [20 | *]  (link new → what was at pos 1)
# current.next = new_node      → [10 | *] → [15 | *]  (link pos 0 → new)
#
# Final: head → [10] → [15] → [20] → [30] → None
#
# COMMON MISTAKE: Doing current.next = new_node BEFORE new_node.next = current.next
# would LOSE the reference to [20] → [30], creating:
#   head → [10] → [15] → None   ← [20] and [30] are now orphaned!

# --- TEST ---
def test_add_at_position():
    ll = LinkedList()
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    
    # Insert in middle
    ll.add_at_position(1, 15)
    assert ll.display() == [10, 15, 20, 30], f"Got {ll.display()}"
    
    # Insert at head via position
    ll.add_at_position(0, 5)
    assert ll.display() == [5, 10, 15, 20, 30]
    
    # Insert at tail via out-of-bounds position
    ll.add_at_position(100, 99)
    assert ll.display() == [5, 10, 15, 20, 30, 99]
    
    print("✅ add_at_position tests passed!")
````

## Cell 6 — delete_at_head


````python
# ============================================================
# SUB-TOPIC: delete_at_head
# Operation: Remove and return the FIRST node's data
# ============================================================

# --- THEORY ---
# What: Removes the node at the front and returns its value.
#
# Why: This + add_at_tail gives you a QUEUE (FIFO).
#      This + add_at_head gives you a STACK (LIFO).
#      Used in: BFS queues, undo stacks, task scheduling.
#
# The trick: Just move head to head.next. The old head becomes
# unreachable (no references to it), and Python's garbage
# collector reclaims its memory automatically.
#
# BEFORE delete_at_head():
#   head → [10 | *] → [20 | *] → [30 | None]
#
# AFTER:
#   head → [20 | *] → [30 | None]
#   (Node 10 has been garbage collected)
#
# Time Complexity: O(1) — no traversal needed
# Space Complexity: O(1)
# Edge Cases:
#   - Empty list: nothing to delete, return None
#   - Single element: head becomes None after deletion

    def delete_at_head(self) -> Optional[int]:
        """
        Remove and return the data of the first node.
        Returns None if the list is empty.
        
        Time: O(1) | Space: O(1)
        
        BEGINNER: We save the data BEFORE moving head, because once we
        do `self.head = self.head.next`, we can't access the old head anymore.
        
        ADVANCED: In languages without garbage collection (C, C++), you'd
        need to explicitly `free(old_head)` to avoid memory leaks. Python
        handles this automatically via reference counting.
        """
        if not self.head:
            return None
        
        data = self.head.data        # Save data before losing reference
        self.head = self.head.next   # Move head forward (old head becomes unreachable)
        self.size -= 1
        return data


# --- DRY RUN TRACE ---
#
# List: head → [10 | *] → [20 | *] → [30 | None], size = 3
#
# delete_at_head():
#   self.head is not None → proceed
#   data = self.head.data = 10
#   self.head = self.head.next → head now points to [20 | *]
#   size = 2
#   return 10
#
# Memory after: [10 | *] node has 0 references → Python frees it
# State: head → [20 | *] → [30 | None]
#
# Edge case — single element:
# List: head → [42 | None], size = 1
#   data = 42
#   self.head = self.head.next = None → head = None (empty list!)
#   size = 0
#   return 42

# --- TEST ---
def test_delete_at_head():
    ll = LinkedList()
    
    # Empty list edge case
    assert ll.delete_at_head() is None
    
    # Single element
    ll.add_at_head(42)
    assert ll.delete_at_head() == 42
    assert ll.head is None
    assert ll.size == 0
    
    # Multiple elements
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    
    assert ll.delete_at_head() == 10
    assert ll.display() == [20, 30]
    assert ll.size == 2
    
    assert ll.delete_at_head() == 20
    assert ll.delete_at_head() == 30
    assert ll.delete_at_head() is None   # Empty again
    
    print("✅ delete_at_head tests passed!")
````

## Cell 7 — delete_at_tail


````python
# ============================================================
# SUB-TOPIC: delete_at_tail
# Operation: Remove and return the LAST node's data
# ============================================================

# --- THEORY ---
# What: Removes the last node from the list.
#
# Why: Useful for implementing deques, removing the oldest item in
#      a fixed-size window, popping from the back of a sequence.
#
# Challenge: To delete the last node, we need to find the
# SECOND-TO-LAST node and set its .next = None.
# That requires traversal: O(n).
#
# BEFORE delete_at_tail():
#   head → [10] → [20] → [30] → None
#
# AFTER:
#   head → [10] → [20] → None
#
# Time Complexity: O(n) — must walk to second-to-last node
# Space Complexity: O(1)
# Edge Cases:
#   - Empty list: return None
#   - Single element: head becomes None

    def delete_at_tail(self) -> Optional[int]:
        """
        Remove and return the data of the last node.
        Returns None if the list is empty.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: We need to stop at the SECOND-TO-LAST node, not the last.
        The condition `while current.next.next` keeps us one step ahead.
        
        ADVANCED: This O(n) cost is why doubly linked lists (DLL) are
        preferred when frequent tail deletions are needed — DLL can do
        this in O(1) using the `prev` pointer of the tail node.
        """
        if not self.head:
            return None
        
        # Special case: only one node
        if not self.head.next:
            data = self.head.data
            self.head = None
            self.size -= 1
            return data
        
        # Walk to second-to-last node
        current = self.head
        while current.next.next:
            current = current.next
        
        data = current.next.data    # Save last node's data
        current.next = None         # Unlink last node
        self.size -= 1
        return data


# --- DRY RUN TRACE ---
#
# List: head → [10] → [20] → [30] → None, size = 3
#
# current = [10]
# Iter 1: current.next.next = [30] (truthy) → current = [20]
# Iter 2: current.next.next = None (falsy) → EXIT loop
#
# current is now [20] (second-to-last)
# data = current.next.data = 30
# current.next = None → [20 | None]
# size = 2
# return 30
#
# State: head → [10] → [20] → None

# --- TEST ---
def test_delete_at_tail():
    ll = LinkedList()
    
    # Empty list
    assert ll.delete_at_tail() is None
    
    # Single element
    ll.add_at_tail(42)
    assert ll.delete_at_tail() == 42
    assert ll.head is None
    
    # Multiple elements
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    
    assert ll.delete_at_tail() == 30
    assert ll.display() == [10, 20]
    assert ll.delete_at_tail() == 20
    assert ll.delete_at_tail() == 10
    assert ll.delete_at_tail() is None   # Empty now
    
    print("✅ delete_at_tail tests passed!")
````

## Cell 8 — delete_by_value


````python
# ============================================================
# SUB-TOPIC: delete_by_value
# Operation: Remove the first occurrence of a given value
# ============================================================

# --- THEORY ---
# What: Search for a value and delete that node if found.
#
# Why: Most real-world deletions are by value, not by index.
#      "Remove student with ID 1042." "Delete task named 'Buy milk'."
#
# The key technique: The "previous pointer" pattern.
#   - Maintain a `prev` pointer one step behind `current`
#   - When you find the target: prev.next = current.next (skip current)
#   - This effectively removes current from the chain
#
# BEFORE delete_by_value(20):
#   head → [10] → [20] → [30] → None
#                  ^
#               target
#
# AFTER:
#   head → [10] → [30] → None
#
# Time Complexity: O(n) — worst case: value at end or not found
# Space Complexity: O(1)
# Edge Cases:
#   - Empty list
#   - Value at head (special case — no prev node!)
#   - Value not in list (return False)
#   - Duplicate values (only delete FIRST occurrence)

    def delete_by_value(self, value: int) -> bool:
        """
        Delete the first node containing the given value.
        Returns True if found and deleted, False otherwise.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: Two-pointer pattern here! `prev` trails one behind `current`.
        When we find the value in current, we do prev.next = current.next
        which "skips over" current.
        
        ADVANCED: The head check is a common pattern. Many LL problems
        have a special case for the head because it has no `prev`. A cleaner
        solution uses a "dummy head" node (sentinel) so ALL nodes have a
        prev, eliminating the special case. See dummy_head pattern below.
        """
        if not self.head:
            return False
        
        # Special case: value is at the head
        if self.head.data == value:
            self.head = self.head.next
            self.size -= 1
            return True
        
        # General case: value is somewhere after head
        prev = self.head
        current = self.head.next
        
        while current:
            if current.data == value:
                prev.next = current.next   # Skip over current
                self.size -= 1
                return True
            prev = current
            current = current.next
        
        return False   # Value not found
    
    # ADVANCED PATTERN: Dummy Head (sentinel) approach
    def delete_by_value_clean(self, value: int) -> bool:
        """
        Cleaner version using a dummy head node (sentinel pattern).
        
        By prepending a fake node before head, ALL real nodes have a `prev`,
        eliminating the head special case.
        
        ADVANCED: The dummy/sentinel pattern appears in many LL problems.
        Trade-off: Slightly more memory (one extra node) for cleaner code.
        """
        dummy = Node(-1)     # Sentinel: fake node before head
        dummy.next = self.head
        prev = dummy
        current = self.head
        
        while current:
            if current.data == value:
                prev.next = current.next
                self.head = dummy.next   # In case we deleted head
                self.size -= 1
                return True
            prev = current
            current = current.next
        
        return False


# --- DRY RUN TRACE ---
#
# List: head → [10] → [20] → [30] → None
# delete_by_value(20):
#
# head.data (10) != 20 → go to general case
# prev = [10], current = [20]
#
# Iteration 1:
#   current.data (20) == 20 → FOUND!
#   prev.next = current.next → [10 | *] → [30 | None]  (skip [20])
#   size -= 1
#   return True
#
# State: head → [10] → [30] → None
#
# delete_by_value(99):
#
# head.data (10) != 99
# prev = [10], current = [30]
# Iter 1: current.data (30) != 99, prev = [30], current = None
# current is None → exit while
# return False

# --- TEST ---
def test_delete_by_value():
    ll = LinkedList()
    
    # Empty list
    assert ll.delete_by_value(5) == False
    
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    ll.add_at_tail(20)   # Duplicate!
    
    # Delete from middle (first occurrence of 20)
    assert ll.delete_by_value(20) == True
    assert ll.display() == [10, 30, 20]   # Only first 20 removed
    
    # Delete from head
    assert ll.delete_by_value(10) == True
    assert ll.display() == [30, 20]
    assert ll.head.data == 30
    
    # Delete from tail
    assert ll.delete_by_value(20) == True
    assert ll.display() == [30]
    
    # Value not found
    assert ll.delete_by_value(999) == False
    
    print("✅ delete_by_value tests passed!")
````

## Cell 9 — search


````python
# ============================================================
# SUB-TOPIC: search
# Operation: Find a value and return its position (0-based)
# ============================================================

# --- THEORY ---
# What: Linear scan through the list to find a value.
#
# Why: Fundamental operation — before updating/deleting by value,
#      you often need to know IF the value exists and WHERE.
#
# Unlike arrays, there's NO random access in linked lists.
# We cannot do `arr[5]` — we MUST start at head and walk.
# This is the core trade-off: arrays are O(1) access, LL is O(n).
#
# Time Complexity: O(n) — worst case: value at end or not found
# Space Complexity: O(1) — only a counter and current pointer
# Edge Cases:
#   - Empty list: return -1
#   - Value not in list: return -1
#   - Multiple occurrences: return index of FIRST occurrence

    def search(self, value: int) -> int:
        """
        Find and return the 0-based index of the first occurrence of value.
        Returns -1 if not found.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: We maintain an index counter as we walk. Each step
        increments the counter. When we find our value, return the counter.
        
        ADVANCED: This is sequential/linear search. For sorted linked lists,
        you can short-circuit early (return -1 as soon as current.data > value),
        but worst-case is still O(n). Binary search requires O(1) index access,
        which LL doesn't have — that's why trees exist.
        """
        current = self.head
        index = 0
        
        while current:
            if current.data == value:
                return index
            current = current.next
            index += 1
        
        return -1   # Not found
    
    def get(self, index: int) -> Optional[int]:
        """
        Return the data at the given 0-based index.
        Returns None if index is out of bounds.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER ANALOGY: Like visiting every page in a book one by one
        to reach page 47 — no random jumping allowed.
        """
        if index < 0 or index >= self.size:
            return None
        
        current = self.head
        for _ in range(index):
            current = current.next
        return current.data


# --- DRY RUN TRACE ---
#
# List: head → [10] → [20] → [30] → None
# search(20):
#
# current = [10], index = 0
# Iter 1: 10 != 20, current = [20], index = 1
# Iter 2: 20 == 20 → return 1
#
# search(99):
# Iter 1: 10 != 99, index = 1
# Iter 2: 20 != 99, index = 2
# Iter 3: 30 != 99, index = 3
# current = None → exit while
# return -1

# --- TEST ---
def test_search():
    ll = LinkedList()
    
    # Empty list
    assert ll.search(5) == -1
    
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    
    assert ll.search(10) == 0   # Head
    assert ll.search(20) == 1   # Middle
    assert ll.search(30) == 2   # Tail
    assert ll.search(99) == -1  # Not found
    
    assert ll.get(0) == 10
    assert ll.get(2) == 30
    assert ll.get(5) is None    # Out of bounds
    
    print("✅ search tests passed!")
````

## Cell 10 — get_size + display


````python
# ============================================================
# SUB-TOPIC: get_size and display
# Operation: Report the list length / visualize the list
# ============================================================

# --- THEORY ---
# get_size():
#   O(1) because we maintain `self.size` counter.
#   Without the counter, we'd need to traverse: O(n).
#   This is a classic space-time trade-off.
#
# display():
#   O(n) — must visit every node to collect values.
#   Used for debugging and testing.
#
# ADVANCED: In production, you'd implement __len__ and __str__
#   so the list works naturally with Python's len() and print().

    def get_size(self) -> int:
        """Return the number of nodes. O(1) due to size counter."""
        return self.size
    
    def is_empty(self) -> bool:
        """Return True if the list has no nodes. O(1)."""
        return self.size == 0
    
    def display(self) -> list[int]:
        """Return all node values as a Python list. O(n)."""
        result = []
        current = self.head
        while current:
            result.append(current.data)
            current = current.next
        return result
    
    def __str__(self) -> str:
        """Human-readable representation: 10 → 20 → 30 → None"""
        nodes = [str(d) for d in self.display()]
        return " → ".join(nodes) + " → None" if nodes else "None"
    
    def __len__(self) -> int:
        """Enables: len(my_list)"""
        return self.size


# --- TEST ---
def test_size_display():
    ll = LinkedList()
    assert ll.get_size() == 0
    assert ll.is_empty() == True
    assert ll.display() == []
    
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    
    assert ll.get_size() == 3
    assert len(ll) == 3
    assert ll.is_empty() == False
    assert ll.display() == [10, 20, 30]
    assert str(ll) == "10 → 20 → 30 → None"
    
    print("✅ get_size / display tests passed!")
````

## Cell 11 — reverse


````python
# ============================================================
# SUB-TOPIC: reverse
# Operation: Reverse the linked list in-place (no extra space)
# ============================================================

# --- THEORY ---
# What: Flip the direction of all .next pointers.
#
# Why: Extremely common interview question. Core technique for:
#   - Palindrome checking
#   - Reversing the second half of a list
#   - K-group reversal problems
#
# The 3-pointer technique (prev, current, next_node):
#   - prev: tracks the "new next" for current node (starts as None)
#   - current: the node we're currently reversing
#   - next_node: saved reference to current.next (so we don't lose it)
#
# BEFORE: head → [1] → [2] → [3] → [4] → None
# AFTER:  head → [4] → [3] → [2] → [1] → None
#
# Time Complexity: O(n) — one pass through the list
# Space Complexity: O(1) — just 3 pointers, no extra data structure
# Edge Cases:
#   - Empty list: nothing to do
#   - Single node: already reversed

    def reverse(self) -> None:
        """
        Reverse the linked list in-place using 3-pointer technique.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: At each step, we flip one arrow. The three pointers
        let us: (1) save where we're going next, (2) flip the current arrow,
        (3) move forward.
        
        ADVANCED: This is the iterative approach. There's also a recursive
        approach that uses the call stack (O(n) space). For very long lists,
        the iterative approach is safer (no stack overflow risk).
        
        The recursive version:
            def reverse_recursive(node):
                if not node or not node.next:
                    return node
                new_head = reverse_recursive(node.next)
                node.next.next = node
                node.next = None
                return new_head
        """
        prev = None
        current = self.head
        
        while current:
            next_node = current.next   # Step 1: Save next (don't lose it!)
            current.next = prev        # Step 2: Flip the arrow
            prev = current             # Step 3: Move prev forward
            current = next_node        # Step 4: Move current forward
        
        self.head = prev               # prev is now the new head


# --- DRY RUN TRACE ---
# List: 1 → 2 → 3 → None
#
# Initial: prev=None, current=[1]
#
# Iteration 1:
#   next_node = [2]
#   current.next = None     → [1] now points to None (arrow flipped!)
#   prev = [1]
#   current = [2]
#   State: None ← [1]   [2] → [3] → None
#
# Iteration 2:
#   next_node = [3]
#   current.next = [1]      → [2] now points to [1]
#   prev = [2]
#   current = [3]
#   State: None ← [1] ← [2]   [3] → None
#
# Iteration 3:
#   next_node = None
#   current.next = [2]      → [3] now points to [2]
#   prev = [3]
#   current = None
#   State: None ← [1] ← [2] ← [3]
#
# current is None → exit while
# self.head = prev = [3]
# Final: head → [3] → [2] → [1] → None ✓

# --- TEST ---
def test_reverse():
    ll = LinkedList()
    
    # Empty list
    ll.reverse()
    assert ll.display() == []
    
    # Single element
    ll.add_at_tail(42)
    ll.reverse()
    assert ll.display() == [42]
    
    # Multiple elements
    ll.add_at_tail(10)
    ll.add_at_tail(20)
    ll.add_at_tail(30)
    # Currently: [42, 10, 20, 30]
    ll.reverse()
    assert ll.display() == [30, 20, 10, 42]
    
    # Reversing twice = original
    ll.reverse()
    assert ll.display() == [42, 10, 20, 30]
    
    print("✅ reverse tests passed!")
````

## Cell 12 — find_middle


````python
# ============================================================
# SUB-TOPIC: find_middle
# Operation: Find the middle node using Floyd's slow/fast pointers
# ============================================================

# --- THEORY ---
# What: Return the middle node of the linked list.
#       For even length, return the second middle (e.g., [1,2,3,4] → node 3).
#
# Why: Used in merge sort on linked lists (split at middle),
#      palindrome checking (reverse second half), finding median.
#
# Naive approach: Count nodes (O(n)), walk to n//2 (O(n)) = 2 passes, O(n)
#
# Smart approach: Floyd's Two-Pointer / Tortoise and Hare
#   - slow moves 1 step at a time
#   - fast moves 2 steps at a time
#   - When fast reaches end, slow is at the middle
#
# WHY DOES THIS WORK?
#   - fast travels 2x as far as slow in the same time
#   - When fast has traveled n steps, slow has traveled n/2 steps
#   - n/2 steps from the start = the middle!
#
# Time Complexity: O(n) — one pass, but single pass!
# Space Complexity: O(1)
# Edge Cases:
#   - Empty list: return None
#   - 1 element: return that element (it's both head and middle)
#   - Even length: returns second middle (common interview convention)

    def find_middle(self) -> Optional[Node]:
        """
        Return the middle node using slow/fast pointer technique.
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: Imagine two runners on a track. The fast one runs twice
        as fast. When the fast one finishes the track, the slow one is
        exactly halfway. Same idea here!
        
        ADVANCED: The condition `while fast and fast.next` handles both
        odd and even length lists correctly.
        - Odd (n=5): fast takes 2,2 steps → at index 4 (end), slow is at 2 (middle ✓)
        - Even (n=4): fast takes 2 steps → at index 2 (fast.next=index3, fast.next.next=None)
          slow is at index 2 (second middle ✓)
        """
        if not self.head:
            return None
        
        slow = self.head
        fast = self.head
        
        while fast and fast.next:
            slow = slow.next         # 1 step
            fast = fast.next.next    # 2 steps
        
        return slow


# --- DRY RUN TRACE ---
# List: 1 → 2 → 3 → 4 → 5 (odd, middle = 3)
#
# slow=1, fast=1
# Iter 1: slow=2, fast=3
# Iter 2: slow=3, fast=5
# Iter 3: fast.next = None → EXIT
# return slow = node(3) ✓
#
# List: 1 → 2 → 3 → 4 (even, second middle = 3)
#
# slow=1, fast=1
# Iter 1: slow=2, fast=3
# Iter 2: slow=3, fast=None → (fast.next = None → EXIT) wait...
#   fast=3, fast.next=4 (truthy) → slow=3, fast=4.next.next... 4.next=None
#   → while fast(4) and fast.next(None) → False → EXIT
# return slow = node(3) ✓

# --- TEST ---
def test_find_middle():
    ll = LinkedList()
    
    # Empty
    assert ll.find_middle() is None
    
    # 1 element
    ll.add_at_tail(1)
    assert ll.find_middle().data == 1
    
    # Odd length: 1 2 3 → middle = 2
    ll.add_at_tail(2)
    ll.add_at_tail(3)
    assert ll.find_middle().data == 2
    
    # Even length: 1 2 3 4 → second middle = 3
    ll.add_at_tail(4)
    assert ll.find_middle().data == 3
    
    # Odd length: 1 2 3 4 5 → middle = 3
    ll.add_at_tail(5)
    assert ll.find_middle().data == 3
    
    print("✅ find_middle tests passed!")
````

## Cell 13 — detect_cycle (Floyd's Cycle Detection)


````python
# ============================================================
# SUB-TOPIC: detect_cycle
# Operation: Detect if the list contains a cycle
# ============================================================

# --- THEORY ---
# What: A cycle exists when a node's .next eventually points BACK
#       to a previous node, creating an infinite loop.
#
# CYCLE EXAMPLE:
#   head → [1] → [2] → [3] → [4]
#                              ↑           ↓
#                             [6] ← [5] ←─┘
#   (traversing this list forever: 1→2→3→4→5→6→4→5→6→...)
#
# Why detect cycles? In practice: corrupted memory, malformed input data,
#   buggy pointer manipulation. In interviews: core LL pattern.
#
# Naive approach: Store visited nodes in a HashSet. O(n) space.
#
# Floyd's Algorithm (Tortoise and Hare):
#   - slow moves 1 step, fast moves 2 steps
#   - If no cycle: fast reaches None
#   - If cycle: fast "laps" slow inside the cycle → they meet!
#
# WHY DO THEY MEET?
#   Once both pointers enter the cycle, the distance between them decreases
#   by 1 each iteration (fast gains 1 step on slow per round). They MUST meet.
#
# Time Complexity: O(n) — at most 2 passes through the list
# Space Complexity: O(1) — only 2 pointers
# Edge Cases:
#   - Empty list: no cycle
#   - Single node pointing to itself: cycle!

    def detect_cycle(self) -> bool:
        """
        Return True if the list has a cycle, False otherwise.
        Uses Floyd's Cycle Detection (Tortoise and Hare).
        
        Time: O(n) | Space: O(1)
        
        BEGINNER: Two runners on a circular track. The fast one eventually
        laps the slow one — they meet. If the track isn't circular (no cycle),
        the fast one just reaches the end.
        
        ADVANCED: Floyd's algorithm can also find the START of the cycle:
        1. Find meeting point (as below)
        2. Reset one pointer to head
        3. Move both at speed 1 → they meet at cycle start
        This is LeetCode 142 (Linked List Cycle II).
        """
        slow = self.head
        fast = self.head
        
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow is fast:       # Same object in memory = cycle!
                return True
        
        return False
    
    def find_cycle_start(self) -> Optional[Node]:
        """
        Return the node where the cycle begins, or None if no cycle.
        ADVANCED: Mathematical proof based on cycle length.
        
        If meeting point is found at distance k from cycle start,
        and cycle start is distance m from head, then:
        head-to-start = start-to-meeting (mod cycle length)
        → Reset slow to head, advance both by 1, they meet at cycle start.
        """
        slow = self.head
        fast = self.head
        
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            if slow is fast:
                break
        else:
            return None   # No cycle
        
        # Reset slow to head, advance both by 1 step
        slow = self.head
        while slow is not fast:
            slow = slow.next
            fast = fast.next
        
        return slow   # Cycle start


# --- DRY RUN TRACE ---
# List with cycle: 1 → 2 → 3 → 4 → 2 (4's next points back to 2)
#
# slow=1, fast=1
# Iter 1: slow=2, fast=3
# Iter 2: slow=3, fast=2 (fast: 3→4→2)
# Iter 3: slow=4, fast=4 (fast: 2→3→4)
# slow is fast → CYCLE DETECTED! return True

# --- TEST ---
def test_detect_cycle():
    # No cycle
    ll = LinkedList()
    ll.add_at_tail(1)
    ll.add_at_tail(2)
    ll.add_at_tail(3)
    assert ll.detect_cycle() == False
    
    # Create a cycle manually: 1 → 2 → 3 → back to node 2
    node1 = Node(1)
    node2 = Node(2)
    node3 = Node(3)
    node1.next = node2
    node2.next = node3
    node3.next = node2   # CYCLE: 3 points back to 2
    
    ll2 = LinkedList()
    ll2.head = node1
    assert ll2.detect_cycle() == True
    
    # Self-loop: node points to itself
    node_self = Node(99)
    node_self.next = node_self
    ll3 = LinkedList()
    ll3.head = node_self
    assert ll3.detect_cycle() == True
    
    print("✅ detect_cycle tests passed!")
````

## Cell 14 — merge_sorted (Merge Two Sorted Lists)


````python
# ============================================================
# SUB-TOPIC: merge_sorted
# Operation: Merge two sorted linked lists into one sorted list
# ============================================================

# --- THEORY ---
# What: Given two sorted linked lists, produce one merged sorted list.
#       (This is the merge step of Merge Sort on linked lists!)
#
# Why: Core algorithm used in external sorting, merge sort, and
#      combining sorted data streams without extra memory allocation.
#
# Approach: Compare heads of both lists, take the smaller one,
#           advance that list's pointer, repeat.
#
# List A: 1 → 3 → 5 → None
# List B: 2 → 4 → 6 → None
# Result: 1 → 2 → 3 → 4 → 5 → 6 → None
#
# DUMMY HEAD TRICK: Use a sentinel node to avoid special-casing
# the first comparison. The dummy's .next is our actual result head.
#
# Time Complexity: O(m + n) — process each node exactly once
# Space Complexity: O(1) — reuse existing nodes, just rewire pointers
#   (Note: recursive version uses O(m+n) call stack space)
# Edge Cases:
#   - Either list is empty: return the other

def merge_sorted_lists(list1: Optional[Node], list2: Optional[Node]) -> Optional[Node]:
    """
    Merge two sorted linked lists and return the merged sorted head.
    Modifies the nodes in-place (rewires .next pointers).
    
    Time: O(m+n) | Space: O(1)
    
    BEGINNER: Two sorted piles of cards. Always pick the smaller top card.
    When one pile runs out, flip the rest of the other pile onto the result.
    
    ADVANCED: The dummy head is a sentinel pattern. We attach it before
    the result list so we always have a `.next` to write to, avoiding
    the first-node special case.
    """
    dummy = Node(-1)    # Sentinel — doesn't become part of final answer
    current = dummy     # "Tail" of our growing result list
    
    while list1 and list2:
        if list1.data <= list2.data:
            current.next = list1    # Attach smaller node to result
            list1 = list1.next      # Advance list1
        else:
            current.next = list2
            list2 = list2.next
        current = current.next      # Advance result tail
    
    # At least one list is exhausted — attach the remaining part
    current.next = list1 if list1 else list2
    
    return dummy.next   # Return actual head (skip the dummy)


# --- DRY RUN TRACE ---
# list1: 1 → 3 → None, list2: 2 → 4 → None
#
# dummy → [d]
# current = dummy
#
# Iter 1: 1 <= 2 → current.next = node(1), list1 = [3], current = node(1)
# Iter 2: 3 > 2  → current.next = node(2), list2 = [4], current = node(2)
# Iter 3: 3 <= 4 → current.next = node(3), list1 = None, current = node(3)
# list1 is None → exit while
# current.next = list2 = [4]
#
# dummy.next = node(1)
# Result: 1 → 2 → 3 → 4 → None ✓

# --- TEST ---
def test_merge_sorted():
    def build(values: list[int]) -> Optional[Node]:
        dummy = Node(-1)
        current = dummy
        for v in values:
            current.next = Node(v)
            current = current.next
        return dummy.next
    
    def to_list(head: Optional[Node]) -> list[int]:
        result, current = [], head
        while current:
            result.append(current.data)
            current = current.next
        return result
    
    # Basic merge
    result = merge_sorted_lists(build([1, 3, 5]), build([2, 4, 6]))
    assert to_list(result) == [1, 2, 3, 4, 5, 6]
    
    # One empty list
    result = merge_sorted_lists(build([1, 2, 3]), None)
    assert to_list(result) == [1, 2, 3]
    
    result = merge_sorted_lists(None, build([4, 5]))
    assert to_list(result) == [4, 5]
    
    # Both empty
    result = merge_sorted_lists(None, None)
    assert result is None
    
    # Duplicates
    result = merge_sorted_lists(build([1, 2, 2]), build([2, 3]))
    assert to_list(result) == [1, 2, 2, 2, 3]
    
    print("✅ merge_sorted tests passed!")

test_merge_sorted()
````

## Cell 15 — COMPLETE CLASS (All Operations Together)


````python
# ============================================================
# COMPLETE SINGLY LINKED LIST — Production-Ready Implementation
# ============================================================
from __future__ import annotations
from typing import Optional

class Node:
    def __init__(self, data: int) -> None:
        self.data: int = data
        self.next: Optional[Node] = None
    def __repr__(self) -> str:
        return f"Node({self.data})"

class LinkedList:
    def __init__(self) -> None:
        self.head: Optional[Node] = None
        self.size: int = 0

    # ── INSERTION ───────────────────────────────
    def add_at_head(self, data: int) -> None:
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        self.size += 1

    def add_at_tail(self, data: int) -> None:
        new_node = Node(data)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
        self.size += 1

    def add_at_position(self, position: int, data: int) -> None:
        if position <= 0:
            self.add_at_head(data)
            return
        if position >= self.size:
            self.add_at_tail(data)
            return
        new_node = Node(data)
        current = self.head
        for _ in range(position - 1):
            current = current.next
        new_node.next = current.next
        current.next = new_node
        self.size += 1

    # ── DELETION ────────────────────────────────
    def delete_at_head(self) -> Optional[int]:
        if not self.head:
            return None
        data = self.head.data
        self.head = self.head.next
        self.size -= 1
        return data

    def delete_at_tail(self) -> Optional[int]:
        if not self.head:
            return None
        if not self.head.next:
            data = self.head.data
            self.head = None
            self.size -= 1
            return data
        current = self.head
        while current.next.next:
            current = current.next
        data = current.next.data
        current.next = None
        self.size -= 1
        return data

    def delete_by_value(self, value: int) -> bool:
        if not self.head:
            return False
        if self.head.data == value:
            self.head = self.head.next
            self.size -= 1
            return True
        prev, current = self.head, self.head.next
        while current:
            if current.data == value:
                prev.next = current.next
                self.size -= 1
                return True
            prev, current = current, current.next
        return False

    # ── SEARCH / ACCESS ─────────────────────────
    def search(self, value: int) -> int:
        current, index = self.head, 0
        while current:
            if current.data == value:
                return index
            current, index = current.next, index + 1
        return -1

    def get(self, index: int) -> Optional[int]:
        if index < 0 or index >= self.size:
            return None
        current = self.head
        for _ in range(index):
            current = current.next
        return current.data

    # ── UTILITY ─────────────────────────────────
    def get_size(self) -> int:
        return self.size

    def is_empty(self) -> bool:
        return self.size == 0

    def display(self) -> list[int]:
        result, current = [], self.head
        while current:
            result.append(current.data)
            current = current.next
        return result

    def reverse(self) -> None:
        prev, current = None, self.head
        while current:
            next_node = current.next
            current.next = prev
            prev, current = current, next_node
        self.head = prev

    def find_middle(self) -> Optional[Node]:
        if not self.head:
            return None
        slow, fast = self.head, self.head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
        return slow

    def detect_cycle(self) -> bool:
        slow, fast = self.head, self.head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
            if slow is fast:
                return True
        return False

    def __str__(self) -> str:
        nodes = [str(d) for d in self.display()]
        return " → ".join(nodes) + " → None" if nodes else "None"

    def __len__(self) -> int:
        return self.size


# ── FULL INTEGRATION TEST ───────────────────────────────────
def test_full():
    ll = LinkedList()
    
    # Build list
    for v in [10, 20, 30, 40, 50]:
        ll.add_at_tail(v)
    assert ll.display() == [10, 20, 30, 40, 50]
    
    # Operations
    ll.add_at_head(5)
    ll.add_at_position(3, 25)
    assert ll.display() == [5, 10, 20, 25, 30, 40, 50]
    
    assert ll.search(25) == 3
    assert ll.search(99) == -1
    assert ll.get(0) == 5
    assert ll.get(6) == 50
    
    ll.delete_at_head()
    ll.delete_at_tail()
    assert ll.display() == [10, 20, 25, 30, 40]
    
    ll.delete_by_value(25)
    assert ll.display() == [10, 20, 30, 40]
    
    assert ll.find_middle().data == 30  # [10,20,30,40] second middle = 30
    
    ll.reverse()
    assert ll.display() == [40, 30, 20, 10]
    
    print("✅ Full integration test passed!")
    print(f"   List: {ll}")

test_full()
````

## Cell 17 — Variation 1: Reverse Linked List (LC 206)


````python
"""
PROBLEM: LeetCode 206 — Reverse Linked List
https://leetcode.com/problems/reverse-linked-list/

Given the head of a singly linked list, reverse the list, and return the reversed list.

Example:
  Input:  1 → 2 → 3 → 4 → 5
  Output: 5 → 4 → 3 → 2 → 1

Constraints: 0 <= n <= 5000, -5000 <= Node.val <= 5000

PATTERN: 3-pointer reversal (same as our reverse() method above)
TIME: O(n) | SPACE: O(1)
FOLLOW-UP: Can you solve it both iteratively and recursively?
"""

# ITERATIVE SOLUTION (O(1) space) — the interview standard answer
def reverseList(head: Optional[Node]) -> Optional[Node]:
    prev, current = None, head
    while current:
        next_node = current.next
        current.next = prev
        prev, current = current, next_node
    return prev

# RECURSIVE SOLUTION (O(n) space — call stack)
def reverseList_recursive(head: Optional[Node]) -> Optional[Node]:
    """
    Base case: empty or single node → return as is
    Recursive case: reverse the rest, then flip the last link
    
    ADVANCED: Recursion goes to the last node first, then unwinds.
    At each unwind step: node.next.next = node (flip), node.next = None (cut old link)
    """
    if not head or not head.next:
        return head
    
    new_head = reverseList_recursive(head.next)  # Reverse rest
    head.next.next = head   # Last node of reversed part points back to current
    head.next = None        # Cut old forward link
    return new_head


# TEST
def build(vals): 
    dummy = Node(0)
    cur = dummy
    for v in vals: cur.next = Node(v); cur = cur.next
    return dummy.next

def to_list(head):
    res = []; cur = head
    while cur: res.append(cur.data); cur = cur.next
    return res

assert to_list(reverseList(build([1,2,3,4,5]))) == [5,4,3,2,1]
assert to_list(reverseList(build([]))) == []
assert to_list(reverseList(build([1]))) == [1]
print("✅ LC 206 passed!")
````

## Cell 18 — Variation 2: Middle of Linked List (LC 876)


````python
"""
PROBLEM: LeetCode 876 — Middle of the Linked List
https://leetcode.com/problems/middle-of-the-linked-list/

Given the head of a singly linked list, return the middle node.
If there are two middle nodes, return the SECOND middle node.

Example 1 (odd):   1 → 2 → [3] → 4 → 5  → return node(3)
Example 2 (even):  1 → 2 → 3 → [4] → 5 → 6  → return node(4)

PATTERN: Slow/fast pointers (Tortoise and Hare)
TIME: O(n) | SPACE: O(1)
"""

def middleNode(head: Optional[Node]) -> Optional[Node]:
    slow, fast = head, head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
    return slow  # slow is at the middle


# TEST
assert middleNode(build([1,2,3,4,5])).data == 3      # Odd: middle
assert middleNode(build([1,2,3,4,5,6])).data == 4    # Even: second middle
assert middleNode(build([1])).data == 1               # Single
assert middleNode(build([1,2])).data == 2             # Two: second middle
print("✅ LC 876 passed!")
````

## Cell 19 — Variation 3: Detect Cycle (LC 141)


````python
"""
PROBLEM: LeetCode 141 — Linked List Cycle
https://leetcode.com/problems/linked-list-cycle/

Given head, determine if the linked list has a cycle.
A cycle exists if a node's next pointer points to a previous node.

PATTERN: Floyd's Cycle Detection (slow/fast pointers)
TIME: O(n) | SPACE: O(1)

BONUS — LC 142: Find cycle start (Floyd's extension)
"""

def hasCycle(head: Optional[Node]) -> bool:
    slow, fast = head, head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False


# TEST
# Build cyclic list: 1 → 2 → 3 → 4 → 2 (back to node 2)
n1, n2, n3, n4 = Node(1), Node(2), Node(3), Node(4)
n1.next = n2; n2.next = n3; n3.next = n4; n4.next = n2

assert hasCycle(n1) == True
assert hasCycle(build([1,2,3])) == False
assert hasCycle(None) == False
print("✅ LC 141 passed!")
````

## Cell 20 — Variation 4: Merge Two Sorted Lists (LC 21)


````python
"""
PROBLEM: LeetCode 21 — Merge Two Sorted Lists
https://leetcode.com/problems/merge-two-sorted-lists/

Merge two sorted linked lists and return it as a sorted list.
The list should be made by splicing together the nodes of the first two lists.

Example:
  Input:  1→2→4  and  1→3→4
  Output: 1→1→2→3→4→4

PATTERN: Dummy head + two-pointer merge
TIME: O(m+n) | SPACE: O(1) — reuses existing nodes
"""

def mergeTwoLists(list1: Optional[Node], list2: Optional[Node]) -> Optional[Node]:
    dummy = Node(-1)
    current = dummy
    
    while list1 and list2:
        if list1.data <= list2.data:
            current.next = list1
            list1 = list1.next
        else:
            current.next = list2
            list2 = list2.next
        current = current.next
    
    current.next = list1 or list2   # Attach remaining (elegant Python idiom)
    return dummy.next


# TEST
result = mergeTwoLists(build([1,2,4]), build([1,3,4]))
assert to_list(result) == [1,1,2,3,4,4]

result = mergeTwoLists(None, build([1,2]))
assert to_list(result) == [1,2]

result = mergeTwoLists(None, None)
assert result is None

print("✅ LC 21 passed!")
````

## Cell 22 — Interview Questions

### 🟢 Easy (Definition / Recall)

**Q: What is the time complexity of inserting at the head vs. tail of a singly linked list without a tail pointer?**

**A:** Head insertion is O(1) — we just redirect the head pointer. Tail insertion is O(n) — we must traverse the entire list to find the last node. This asymmetry is why stacks (LIFO, head operations) are faster on linked lists than queues (FIFO, head+tail) unless a tail pointer is maintained.

---

**Q: What does `current.next.next` mean, and when can it cause an error?**

**A:** It means "follow current to the next node, then follow that node's next pointer." It causes a NoneType AttributeError when `current.next` is None — you're trying to access `.next` on None. Always guard with `while current.next and current.next.next` when using look-ahead.

---

### 🟡 Medium (Application / Implementation)

**Q: How would you find the kth node from the END of a linked list in a single pass?**

**A:** Two-pointer technique. Advance the `fast` pointer k steps ahead of `slow`. Then move both at the same pace until `fast` reaches the end. At that point, `slow` is exactly k steps from the end.


````python
def kth_from_end(head: Optional[Node], k: int) -> Optional[Node]:
    slow, fast = head, head
    # Advance fast by k steps
    for _ in range(k):
        if not fast:
            return None  # k > list length
        fast = fast.next
    # Move both until fast hits end
    while fast:
        slow = slow.next
        fast = fast.next
    return slow
````


---

**Q: How would you check if a linked list is a palindrome in O(n) time and O(1) space?**

**A:** Three steps:
1. Find the middle (slow/fast pointers)
2. Reverse the second half of the list in-place
3. Compare the first half with the reversed second half

If all values match, it's a palindrome. Optionally, restore the list by reversing again.

---

### 🔴 Hard (Architecture / Trade-off)

**Q: Why does LRU Cache use a doubly linked list instead of a singly linked list?**

**A:** LRU Cache requires O(1) deletion from an ARBITRARY position. A doubly linked list gives each node a `prev` pointer, so deletion is:

```
node.prev.next = node.next
node.next.prev = node.prev
```

With a singly linked list, to delete a node, you need its previous node, requiring O(n) traversal to find it. The DLL + HashMap combination gives:
- HashMap: O(1) lookup of any node by key
- DLL: O(1) delete (using prev pointer) and O(1) move-to-front

This is a real engineering trade-off: 2x memory per node (extra `prev` pointer) for O(1) vs O(n) deletion.

---

**Q: You're building a message queue system processing millions of messages per second. Would you use a linked list or a circular buffer (array)? Justify with complexity.**

**A:** Circular buffer (array) for this use case, because:

1. **Cache locality:** Array elements are contiguous in memory → CPU cache hits → dramatically better throughput at millions of ops/sec
2. **Memory efficiency:** No per-element pointer overhead (8 bytes/element vs 56 bytes/node in Python)
3. **Predictable latency:** No pointer chasing → consistent memory access patterns

Linked list wins when: message sizes are huge and unpredictable, or when arbitrary insertion/deletion priority queuing is needed.

The data structure choice here has **real production impact**: Redis (in-memory DB) uses a **ziplist** (array-like) for small lists and only switches to a doubly linked list (quicklist) for large ones — exactly this trade-off.

║ add_at_tail            ║ Walk to current.next = None  ║ O(n)    ║ O(1)    ║ Empty list (head = new)       ║
║ delete_at_head         ║ head = head.next             ║ O(1)    ║ O(1)    ║ Empty, single element         ║
║ delete_at_tail         ║ Walk to current.next.next    ║ O(n)    ║ O(1)    ║ Empty, single (head = None)   ║
║ delete_by_value        ║ prev.next = current.next     ║ O(n)    ║ O(1)    ║ Value at head, not found      ║
║ search                 ║ Linear scan with index count ║ O(n)    ║ O(1)    ║ Empty, not found              ║
║ reverse                ║ 3 ptrs: prev, curr, next_n   ║ O(n)    ║ O(1)    ║ Empty, single node            ║
║ find_middle            ║ Slow(1) + Fast(2) pointers   ║ O(n)    ║ O(1)    ║ Even→second middle            ║
║ detect_cycle           ║ Floyd: slow==fast → cycle    ║ O(n)    ║ O(1)    ║ Self-loop, tail→head          ║
║ merge_sorted           ║ Dummy head + compare heads   ║ O(m+n)  ║ O(1)    ║ Either list empty             ║
╚════════════════════════╩══════════════════════════════╩═════════╩═════════╩═══════════════════════════════╝

KEY PATTERNS TO REMEMBER:
  1. Dummy Head      — Eliminates head-special-case in deletion/merge
  2. Slow/Fast Ptr   — Middle, cycle detection, kth-from-end
  3. Prev Pointer    — Required for all deletion operations
  4. Save Next First — ALWAYS save next_node before modifying .next
  5. Floyd's Math    — Meeting point → reset one pointer → cycle start
